## Revision

    lists: []
    tuples: () or data,data2,data3
    set : {} and contains comma separated values only 
    dict : {key:value,key2:value2} : key-value pairs
    boolean : True and False
    string : whatever information is enclosed in quotes(",',''')


```python
print("1",type([12,13,14])) # list
print("2",type((12,13,14))) # tuple
print("3",type("sindhura")) # string
print("4",type({"test","sample"})) # set
print("5",type({1:101,2:202})) # dict
print("6",type(True)) # Boolean
print("7",type("True")) # string
#example:  1 string
```
