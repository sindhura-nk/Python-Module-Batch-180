**Consider employee salaries corresponding to different departments**

    Category the salaries according to below criteria:
    a) salaries <= 30000 : low
    b) 30000<salaries<=50000 : mid
    c) salaries>50000 : high

Display the final list that categories salaries into [low,mid,high]

```python
sal = [50000,70000,38000,19000]
# [mid,high,mid,low]
 
# Create an empty list that stores salary groups
sal_groups = []
# Run a loop on salaries
for salary in sal:
    if salary<=30000: # Condition1
        sal_groups.append("Low")
    elif 30000<salary<=50000: # Condition2
        sal_groups.append("Mid")
    elif salary>50000: # COndition3
        sal_groups.append("High")
    else: # Logic if all conditions failed
        sal_groups.append("Invalid Data")

# Come out of the loop
print(sal)
print(sal_groups)
```

**Create a list that contains only even numbers from a range of numbers**

consider range of numbers between 560009 to 678456

    even numbers : 2,4,6,8,10
    odd numbers : 1,3,5,7,9,11

even numbers are all divisible by 2 and the remainder will be 0

```python
# Define an empty new list
even_nums = []
# Intialize i to 560009
i = 560009
# condition in while
while i<=678456:
    if i%2==0:
        even_nums.append(i)
    i+=1
```
