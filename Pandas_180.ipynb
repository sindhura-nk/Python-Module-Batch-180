**Consider startup from either New York or California whose investment in ADMIN is more than 100000**

combination of OR and AND logical conditions

OR is between New York and california

AND is between the above result and ADMIN>100000

```python
fil4 = (df['STATE']=='California') | (df['STATE']=='New York')
fil5 = df['ADMIN']>100000

# display output only for ADMIN and STATE
df[fil4 & fil5][['ADMIN','STATE']]
```

**Sort**

syntax: asc: df.sort_values(by=['column_name'])

desc:  df.sort_values(by=['column_name'],ascending=False)

Ascending or Descending order

1) Numerical data: smallest to largest or vice versa
2) Text data: ascending: a-z or descending: z-a

**Business Scenario: Consider New York state and check for startups whose investment in MKT is less 150000.**

    a) Sort this dataframe in ascending order of PROFIT columns(least to highest).
    b) Sort the original frame in ascending order of STATE column (a-z)

```python
df_ny_mkt = df.query("(STATE=='New York')&(MKT<=150000)")
df_ny_mkt
```

```python
df_ny_mkt.sort_values(by='PROFIT')
```
